```python
first = {2, 1, 7, 0, 9}
second = {2, 4, 6, 1}
new_set = set()

for num in first:
  if num not in second:
    new_set.add(num)

for num in second:
  if num not in first:
    new_set.add(num)

new_set
```

```python
n = 5
k = 3
lst = []

for i in range(k):
  lst.append(n//k)

if n % k != 0:
  for i in range(len(lst)-1):
    lst[i] += 1
    if sum(lst) == n:
      break

lst
```
