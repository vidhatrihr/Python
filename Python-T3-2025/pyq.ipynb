```python
delta = 5

"="*delta + ">"

delta = -5

"<" + "="*abs(delta)
```

```python
deltas = [1, 4, -2, 5, 0, -3, 0, -3]
pos = 1

for delta in deltas:
  if delta == 0:
    " "*(pos-1) + "="
    continue

  if delta > 0:
    " "*(pos-1) + "="*delta + ">"
  else:

    " "*(pos+delta-1) + "<" + "="*abs(delta)

  pos += delta
```

```python
word = "zebra"
result = "zbare"

even_char = ""
odd_char = ""
for i, char in enumerate(word):
  if i % 2 == 0:
    even_char += char
  else:
    odd_char += char
odd_char = odd_char[::-1]

even_char + odd_char == result
```
