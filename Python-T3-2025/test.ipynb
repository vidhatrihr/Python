```python
L = [90, 47, 8, 18, 10, 7]
S = [L[0]]  # list containing just one element
for i in range(1, len(L)):
  flag = True
  for j in range(len(S)):
    if L[i] < S[j]:
      before_j = S[: j]  # elements in S before index j
      new_j = [L[i]]		# list containing just one element
      after_j = S[j:]  # elements in S starting from index j
      # what is the size of S now?
      S = before_j + new_j + after_j
      # what is the size of S now?
      flag = False
      break
  if flag:
    S.append(L[i])
print(S)
```

```python
def is_greater_than_5(numbers: list) -> list:
  '''Given a list of numbers, return a list of bools corresponding to whether the number is greater than 5'''
  return [x > 5 for x in numbers]
```

```python
def indices_of_big_words(words) -> list:
  '''Given a list of words, find the indices of the big words(length greater than 5).
  '''
  big_words = []
  lengths = list(map(lambda word: len(word), words))
  is_more_than_5 = list(zip(lengths, is_greater_than_5(lengths)))

  print(is_more_than_5)
  for value in enumerate(is_more_than_5):
    if value[1]:
      print(type(value), value[1][1])
  return big_words


indices_of_big_words(["Apple", "Banana", "Orange", "Kiwi", "Cherry"])
```

```python
import itertools
```

```python
def groupby(data: list, key: callable):
  '''
  Given a list of items, and a key, create a dictionary with the key as key function called 
  on item and the list of items with the same key as the corresponding value. 
  The order of items in the group should be the same order in the original list
  '''

  grouped_dict = dict(
      map(lambda item: (item[0], list(item[1])),
          itertools.groupby(data, key))
  )
  return grouped_dict


fruits = ["Apple", "Banana", "Avocado", "Amla", "Black berry", "Blue berry"]

groupby(fruits, lambda x: x[0])

# {
#     "A": ["Apple", "Avocado", "Amla"],
#     "B": ["Banana", "Black berry", "Blue berry"]
# }
```

```python
from itertools import groupby

fruits = ['apple', 'banana', 'apricot']

sorted_fruits = sorted(fruits)
grouped = groupby(sorted_fruits, (lambda x: x[0]))

dictionary = {}
for key, value in grouped:
  dictionary[key] = list(value)

dictionary
```
