```python
def factorial(num):
  factorial = 1
  for x in range(1, num+1):
    factorial *= x
  if num == 0:
    return 1
  else:
    return factorial


factorial(3)
factorial(5)
factorial(0)
factorial(7)
```

```python
def check_leap_year(year):
  if year % 100 == 0:
    if year % 400 == 0:
      return True

  elif year % 4 == 0:
    return True

  return False


check_leap_year(2020)
check_leap_year(9999)
check_leap_year(1956)
check_leap_year(2000)
check_leap_year(2100)
```
