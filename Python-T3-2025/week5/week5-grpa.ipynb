```python
def get_freqs(sentence):
  from string import ascii_lowercase

  sentence = sentence.lower()
  freq = {}
  for char in sentence:
    if char not in ascii_lowercase:
      continue

    if char not in freq:
      freq[char] = 0
    freq[char] += 1

  return freq
```

```python
word = 'Anagram'
word = word.lower()

dict_freq = get_freqs(word)
dict_freq
max_freq = max(dict_freq.values())
max_freq
```

```python
sent = 'vidu like anagrams and apples'

words = sent.split(' ')

best_word = None
for word in words:
  if not best_word:
    best_word = word
    continue
  if max(get_freqs(word).values()) > max(get_freqs(best_word).values()):
    best_word = word

best_word.strip()
```
